##**2.2 Aplicação de Filtros**


```python
img1= cv2.imread('banco1.jpg')
img_rgb1= cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
```

```python
media_3 = cv2.blur(img_rgb1, (3,3))
media_5 = cv2.blur(img_rgb1, (5,5))

fig, ax = plt.subplots(1,3, figsize=(8,5))

ax[0].imshow(img_rgb1)
ax[0].set_title("Original")

ax[1].imshow(media_3)
ax[1].set_title("Média 3x3")

ax[2].imshow(media_5)
ax[2].set_title("Média 5x5")

for a in ax:
    a.axis('off')

plt.show()
```

```python
gauss_1 = cv2.GaussianBlur(img_rgb1, (5,5), 1)
gauss_3 = cv2.GaussianBlur(img_rgb1, (5,5), 3)

fig, ax = plt.subplots(1,3, figsize=(8,5))

ax[0].imshow(img_rgb1)
ax[0].set_title("Original")

ax[1].imshow(gauss_1)
ax[1].set_title("Gaussiano σ=1")

ax[2].imshow(gauss_3)
ax[2].set_title("Gaussiano σ=3")

for a in ax:
    a.axis('off')

plt.show()
```

```python
mediana_3 = cv2.medianBlur(img_rgb1, 3)
mediana_5 = cv2.medianBlur(img_rgb1, 15)

fig, ax = plt.subplots(1,3, figsize=(8,5))

ax[0].imshow(img_rgb1)
ax[0].set_title("Original")

ax[1].imshow(mediana_3)
ax[1].set_title("Mediana 3x3")

ax[2].imshow(mediana_5)
ax[2].set_title("Mediana 15x15")

for a in ax:
    a.axis('off')

plt.show()
```

## 3.1 - Detecção de bordas sem suavização

```python
img_gray1= cv2.imread('banco1.jpg')
banco1_gray = cv2.cvtColor(img_gray1, cv2.COLOR_BGR2GRAY)
```

```python
sobelx_banco1= cv2.Sobel(banco1_gray, cv2.CV_64F, 1, 0, ksize=5)
sobely_banco1= cv2.Sobel(banco1_gray, cv2.CV_64F, 0, 1, ksize=5)

magnitude_banco1 = cv2.magnitude(sobelx_banco1, sobely_banco1)

fig, ax = plt.subplots(1,3, figsize=(8,5))

ax[0].imshow(sobelx_banco1, cmap='gray')
ax[0].set_title("Sobel X")

ax[1].imshow(sobely_banco1, cmap='gray')
ax[1].set_title("Sobel Y")

ax[2].imshow(magnitude_banco1, cmap='gray')
ax[2].set_title("Magnitude")

for a in ax:
    a.axis('off')

plt.show()
```

```python
kernelx_banco1 = np.array([[1,0,-1],
                    [1,0,-1],
                    [1,0,-1]])

kernely_banco1 = np.array([[1,1,1],
                    [0,0,0],
                    [-1,-1,-1]])

prewittx_banco1 = cv2.filter2D(banco1_gray, -1, kernelx_banco1)
prewitty_banco1 = cv2.filter2D(banco1_gray, -1, kernely_banco1)

prewitt_mag_banco1 = cv2.magnitude(np.float32(prewittx_banco1), np.float32(prewitty_banco1))

fig, ax = plt.subplots(1,3, figsize=(8,5))

ax[0].imshow(prewittx_banco1, cmap='gray')
ax[0].set_title("Prewitt X")

ax[1].imshow(prewitty_banco1, cmap='gray')
ax[1].set_title("Prewitt Y")

ax[2].imshow(prewitt_mag_banco1, cmap='gray')
ax[2].set_title("Magnitude")

for a in ax:
    a.axis('off')

plt.show()
```

```python
canny1_banco1 = cv2.Canny(banco1_gray, 50, 100)
canny2_banco1= cv2.Canny(banco1_gray, 100, 200)

fig, ax = plt.subplots(1,2, figsize=(5,5))

ax[0].imshow(canny1_banco1, cmap='gray')
ax[0].set_title("Canny 50-100")

ax[1].imshow(canny2_banco1, cmap='gray')
ax[1].set_title("Canny 100-200")

for a in ax:
    a.axis('off')

plt.show()
```

### **4.2 Aplicação dos filtros após suavizar**
